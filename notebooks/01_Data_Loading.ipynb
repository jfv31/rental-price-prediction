# 01 – Data Loading

Einlesen der Rohdaten (ca. 9,5 GB) per Lazy Scan, Basis-Filter `mietekalt > 0`, Ablage als Parquet.

```python
import polars as pl
import os

RAW_PATH = r"C:\Users\Admin\Desktop\Übertrag\Rental-price-prediction\data\raw\WM_SUF_mitText.csv"
PROCESSED_DIR = r"C:\Users\Admin\Desktop\Übertrag\Rental-price-prediction\data\processed"
FULL_PATH = os.path.join(PROCESSED_DIR, "AUTO_WM_Full.parquet")

print(f"Lese Rohdaten: {RAW_PATH}")
```

```python
# Lazy Scan der Rohdaten
df_lazy = pl.scan_csv(
    RAW_PATH,
    infer_schema_length=10000,
    schema_overrides={
        "ergg_1km": pl.String,
        "mietekaution": pl.String,
        "courtage": pl.String,
        "freiab": pl.String
    }
)

# Basis-Filter (Miete > 0)
df_full = (
    df_lazy
    .filter(pl.col("mietekalt") > 0)
    .collect()
)

os.makedirs(PROCESSED_DIR, exist_ok=True)
df_full.write_parquet(FULL_PATH)

print(f"{df_full.height} Zeilen gespeichert unter {FULL_PATH}")
```
